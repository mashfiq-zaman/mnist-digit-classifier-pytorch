```python
w=torch.tensor(3.0,requires_grad=True)
x=torch.tensor(2.0)
y_pred=w*x
y_true=torch.tensor(10.0)
loss=((y_pred-y_true)**2)
print(loss)
loss.backward()
#print(w.grad)
print("prediction", y_pred.item())
print("Loss", loss.item())
print("w.grad", w.grad.item()) 
```

```python
#input
x=torch.tensor(2.0)

#learnable num
w= torch.tensor(3.0, requires_grad=True)
#correct ans
y_true=torch.tensor(10.0)

lrate=0.1

y_pred=w*x
loss=((y_pred-y_true)**2)

loss.backward()
print("before update")
print("W:",w.item())
print("Loss", loss.item())
print("gradient", w.grad.item())  

with torch.no_grad(): #donot track
    w-=lrate*w.grad
w.grad.zero_()

print("after update")
print("W:",w.item())
 
```

```python
x=torch.tensor([1.,2.,3.,4.])
y_true=torch.tensor([5.,10.,15.,20.])

w=torch.tensor(0.0, requires_grad=True)

lr=0.1
epochs=15

for epoch in range(epochs):
    #prediction
    y_pred=w*x

    #loss
    loss=((y_pred-y_true)**2).mean()

    #find the mistake
    loss.backward()

    with torch.no_grad(): #when updating
        w-=lr*w.grad
    w.grad.zero_() 
    print(f"epoch {epoch+1}: w={w.item():.4f},loss={loss.item():.4f}")

```

```python
x = torch.tensor([1., 2., 3., 4., 5., 6.])
y_true =torch.tensor([3., 8., 13., 18., 23., 28.]) 

w=torch.tensor(0.0, requires_grad=True)
b=torch.tensor(0.0, requires_grad=True)

lr=0.01
epochs=1000
for i in range( epochs):
    y_pred=x*w-b
    loss=((y_pred-y_true)**2).mean()
    loss.backward()

    with torch.no_grad():
       w-=lr*w.grad
       b-=lr*b.grad
    w.grad.zero_()
    b.grad.zero_()
    print(f"epoch {i+1}: w={w.item():.4f},loss={loss.item():.4f}")
    print(f"epoch {i+1}: b={b.item():.4f},loss={loss.item():.4f}")
```
